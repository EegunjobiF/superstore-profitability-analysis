## Introduction

Reliable analysis depends on having clean, consistent, and well-structured data. This notebook focuses on preparing the **Superstore dataset** for exploratory and business analysis.

The primary objective is to **inspect, clean, validate, and standardize the dataset** while preserving the underlying meaning of the data. The cleaning process includes identifying missing values, investigating whether missingness is random or structural, standardizing formats and categorical values, and validating data consistency.

Rather than automatically modifying every irregularity, each data-quality issue is investigated and a decision is made based on the evidence available in the dataset.

The cleaned and validated dataset produced from this notebook will serve as the foundation for the subsequent **Exploratory Data Analysis (EDA) and profitability analysis**.


```python
#Import the necessary libraries
import pandas as pd
```

```python
# Load source dataset
df = pd.read_csv('ecommerce_sales_customer_analytics_150k.csv')
```

Data audit and cleaning

```python
# 1. Show all columns instead of truncating with...
pd.set_option('display.max_columns', None)

# 2. Get dataset size
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
print("-" * 50) # just a separator line

# 3. Preview the first 5 rows with all columns visible
df.head()
```

```python
df.describe() #Describing the data 
```

```python
# Standardize order date
df['order_date'] = pd.to_datetime(
    df['order_date'],
    format='%Y-%m-%d'
)

# Standardize order time
df['order_time'] = pd.to_datetime(
    df['order_time'],
    format='%H:%M:%S'
).dt.time
```

```python
# Verify the data types after standardizing date and time
df[['order_date', 'order_time']].dtypes
```

```python
# Sort orders chronologically by order date
df.sort_values(
    by='order_date',
    inplace=True,
    ascending=True
)
```

```python
# Count unique values in each column and sort from least to most
df.nunique().sort_values()
```

```python
# Inspect unique values in categorical columns
categorical_columns = [
    'gender',
    'customer_segment',
    'customer_type',
    'region',
    'order_status',
    'sales_channel',
    'payment_method',
    'payment_status',
    'currency',
    'shipping_method',
    'warehouse',
    'delivery_status',
    'return_status',
    'return_reason',
    'review_sentiment',
    'marketing_channel',
    'is_repeat_customer'
]

for column in categorical_columns:
    print(f"\n{column}:")
    print(df[column].unique())
```

```python
# Summarize missing values by count and percentage
null_summary = pd.DataFrame({
    'Null Count': df.isnull().sum(),
    'Null Percentage': (df.isnull().sum() / len(df)) * 100
}).sort_values(by='Null Count', ascending=False)

null_summary
```

```python
# Check order status for returned orders
df.loc[
    df['return_status'] == 'Returned',
    'order_status'
].value_counts(dropna=False)
```

```python
# Check order status for missing return statuses
df.loc[
    df['return_status'].isna(),
    'order_status'
].value_counts()
```

```python
# Check return reasons for orders with missing return status
df.loc[
    df['return_status'].isna(),
    'return_reason'
].value_counts(dropna=False)
```

```python
# Check order status for orders with a return reason
df.loc[
    df['return_reason'].notna(),
    'order_status'
].value_counts(dropna=False)
```

```python
# Replace structural missing values with meaningful labels for non-returned orders
df['return_status'] = df['return_status'].fillna('Not Returned')
df['return_reason'] = df['return_reason'].fillna('No Return')
```

The missing values were identified as structural missingness because they represent orders with no return. Therefore, they were replaced with meaningful labels (Not Returned and No Return) to make the data easier to interpret during analysis.

```python
# Confirm that no null values remain in the return fields
df[['return_status', 'return_reason']].isnull().sum()
```

```python
# Compare discount amounts for orders with and without a coupon code
df.groupby(
    df['coupon_code'].isna()
)['discount_amount'].describe()
```

```python
# Check discount values for orders with missing coupon codes
df.loc[
    df['coupon_code'].isna(),
    'discount_amount'
].value_counts()
```

```python
# Check campaign names for orders with missing coupon codes
df.loc[
    df['coupon_code'].isna(),
    'campaign_name'
].value_counts(dropna=False)
```

```python
# Check discount amounts for orders with a coupon code
df.loc[
    df['coupon_code'].notna(),
    ['coupon_code', 'discount_amount']
].groupby('coupon_code')['discount_amount'].agg(['count', 'mean', 'min', 'max']).sort_values('count', ascending=False)
```

```python
# Compare whether discounts were applied to orders with missing coupon codes
df.groupby(
    df['coupon_code'].isna()
)['discount_amount'].agg(['count', 'sum', 'mean', 'min', 'max'])
```

Missing coupon_code values were retained as NaN because the available data does not provide sufficient evidence to determine the actual coupon code. Validation showed that orders with missing coupon codes can still receive substantial discounts and may be associated with marketing campaigns. Therefore, replacing the missing values with a label such as "No Coupon" could introduce inaccurate information.

```python
# Check campaign names and missing values
df['campaign_name'].value_counts(dropna=False)
```

```python
# Compare discounts across campaign names and missing values
df.groupby('campaign_name', dropna=False)['discount_amount'].agg(['count', 'mean', 'sum'])
```

Missing campaign_name values were retained as NaN because the available data does not provide sufficient evidence that missing campaign names represent the absence of a campaign. Validation showed that records with missing campaign names still have substantial discount activity. Therefore, replacing missing values with "No Campaign" could introduce inaccurate information.

```python
# Check delivery status specifically for missing estimated delivery days
df[df['estimated_delivery_days'].isna()]['delivery_status'].value_counts(dropna=False)
```

Missing estimated_delivery_days values were retained as NaN because all 24,557 missing values correspond to cancelled orders. Since cancelled orders did not proceed to delivery, an estimated delivery time is not applicable. Therefore, the missing values were not imputed.
